# AI vs ML vs Deep Learning

## Introduction

Artificial Intelligence (AI), Machine Learning (ML), and Deep Learning (DL) are important concepts in modern technology. They help computers perform tasks, learn from data, and recognize patterns.

This project explains their definitions, differences, real-world examples, and relationships.


## 1. Definitions

### Artificial Intelligence (AI)

Artificial Intelligence is a technology that helps computers perform tasks that usually require human intelligence. It can help machines solve problems, understand language, and make decisions.

**Example:** Chatbots and virtual assistants.

### Machine Learning (ML)

Machine Learning is a part of AI that helps computers learn from data and make predictions without being programmed for every situation.

**Example:** Netflix suggesting movies based on our interests.

### Deep Learning (DL)

Deep Learning is a part of Machine Learning that uses neural networks with multiple layers to learn complex patterns from data.

**Example:** Face recognition in mobile phones.
